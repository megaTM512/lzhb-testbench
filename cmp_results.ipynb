```python
import pandas
import numpy as np
```

```python
normal_csv = pandas.read_csv("testbench_results.csv")
packed_csv = pandas.read_csv("testbench_results_packed.csv")
```

```python
# Error in testbench_results: 
# Error in benchmark: We divided by batchsize twice
normal_csv["AVERAGE_ACCESS_CHAR_TIME"] = normal_csv["AVERAGE_ACCESS_CHAR_TIME"] * normal_csv["BATCH_SIZE"]
```

```python
# Calculate the difference in AVERAGE_ACCESS_CHAR_TIME
comparison_df = pandas.DataFrame({
  'FILE_NAME': normal_csv['FILE_NAME'],
  'MAX_HEIGHT': normal_csv['MAX_HEIGHT'],
  'PHRASE_NUM': normal_csv['PHRASE_NUM'],
  'Normal_Access_Time': normal_csv['AVERAGE_ACCESS_CHAR_TIME'],
  'Packed_Access_Time': packed_csv['AVERAGE_ACCESS_CHAR_TIME'],
  'Difference': packed_csv['AVERAGE_ACCESS_CHAR_TIME'] - normal_csv['AVERAGE_ACCESS_CHAR_TIME'],
  'Percent_Change': ((packed_csv['AVERAGE_ACCESS_CHAR_TIME'] - normal_csv['AVERAGE_ACCESS_CHAR_TIME']) / normal_csv['AVERAGE_ACCESS_CHAR_TIME'] * 100)
})

print(f"Average difference: {comparison_df['Difference'].mean():.6f} ns")
print(f"Average percent change: {comparison_df['Percent_Change'].mean():.2f}%")
print(f"Median percent change: {comparison_df['Percent_Change'].median():.2f}%")
print(f"\nPacked version is slower in {(comparison_df['Difference'] > 0).sum()} out of {len(comparison_df)} cases")
print(f"\nSummary statistics:")
print(comparison_df[['Difference', 'Percent_Change']].describe())

comparison_df.head()
```

```python
# Calculate space savings comparison
space_df = pandas.DataFrame({
  'FILE_NAME': normal_csv['FILE_NAME'],
  'MAX_HEIGHT': normal_csv['MAX_HEIGHT'],
  'PHRASE_NUM': normal_csv['PHRASE_NUM'],
  'Normal_Bytes': normal_csv['TOTAL_ACCESS_CHARS'],
  'Packed_Bytes': packed_csv['BYTES'],
  'Bytes_Saved': normal_csv['TOTAL_ACCESS_CHARS'] - packed_csv['BYTES'],
  'Percent_Saved': ((normal_csv['TOTAL_ACCESS_CHARS'] - packed_csv['BYTES']) / normal_csv['TOTAL_ACCESS_CHARS'] * 100)
})

print(f"Average bytes saved: {space_df['Bytes_Saved'].mean():.0f} bytes")
print(f"Average percent saved: {space_df['Percent_Saved'].mean():.2f}%")
print(f"Median percent saved: {space_df['Percent_Saved'].median():.2f}%")
print(f"\nPacked version saves space in {(space_df['Bytes_Saved'] > 0).sum()} out of {len(space_df)} cases")
print(f"\nSummary statistics:")
print(space_df[['Bytes_Saved', 'Percent_Saved']].describe())

space_df.head()
```
